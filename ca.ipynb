Aquí tienes la lista exacta de lo que te puede preguntar y lo que debes tener ya preparado para no quedar pillada:

1. Parámetros del generador
Debes poder responder sin dudar:
- tamaño del mapa
- proporción de cada tipo de terreno
- densidad de obstáculos
- número de POIs
- número de zonas de comunicación
- semilla
- qué hace el generador si la instancia no es viable

Qué debes tener:
- una función `generar_mapa(...)` con parámetros explícitos
- una política clara: “si la instancia no tiene solución, se regenera; si falla varias veces, se descarta”

2. Garantía de solución
La profesora puede preguntar:
- “¿Cómo sabes que existe camino?”
- “¿Qué pasa si un POI queda aislado?”
- “¿Qué haces si hay una instancia imposible?”

Qué debes tener:
- comprobación de conectividad
- regeneración automática
- explicación de descarte/validación
- mención en el informe de que se evita instancias infactibles

3. Batería y recarga
Pregunta típica:
- “¿Dónde se recarga?”
- “¿Qué consume energía?”
- “¿Qué pasa si la batería se agota?”

Qué debes tener:
- energía por terreno
- recarga en base / zona / estaciones
- regla clara de cuándo recargar
- consumo de movimiento y análisis

4. Memoria
Pregunta típica:
- “¿Qué ocupa memoria?”
- “¿Cuándo se libera?”
- “¿Qué pasa si supera el límite?”

Qué debes tener:
- cada POI tiene datos asociados
- análisis y transmisión consumen memoria
- la memoria se libera tras procesar/transmitir
- límite de memoria bien definido

5. Comparación de búsqueda
Pregunta típica:
- “¿Solo haces A*?”
- “¿Qué pasa con BFS o Dijkstra?”
- “¿Qué métricas mides?”
- “¿Manhattan es admisible?”

Qué debes tener:
- BFS
- Dijkstra/coste uniforme
- A*
- métricas: coste, nodos expandidos, tiempo
- explicación de que Manhattan es admisible si el coste mínimo es 1

6. Modelado PDDL
Pregunta típica:
- “¿Qué modela el dominio?”
- “¿Cómo representas batería y memoria?”
- “¿pyperplan puede manejar números?”

Qué debes tener:
- dominio adaptado al rover
- estado de batería/memoria discretizado
- predicados y acciones bien explicados
- explicación de qué hace el planner y qué hace la búsqueda

7. Optimización del orden de POIs
Pregunta típica:
- “¿Qué es la decisión difícil?”
- “¿Qué metaheurística usas?”
- “¿Cómo representas una solución?”
- “¿Qué función objetivo evalúa?”

Qué debes tener:
- orden de visita como permutación
- solución representada por secuencia
- vecino = intercambio/inversión
- objetivo = coste + penalización + recompensa
- al menos dos metaheurísticas comparadas

8. Integración completa
Pregunta típica:
- “¿El planner genera el plan y luego la búsqueda resuelve cada movimiento?”
- “¿La semilla crea todo el problema?”
- “¿Se ejecuta en cadena?”

Qué debes tener:
- semilla -> mapa -> POIs -> plan -> A* para cada movimiento -> evaluación final
- explicación de la separación de capas

9. Evaluación experimental
Pregunta típica:
- “¿Cuántas instancias?”
- “¿Qué medias y dispersión?”
- “¿Qué comparas?”
- “¿Qué pasa si aumentan los POIs?”

Qué debes tener:
- varias semillas
- varios tamaños o dificultades
- media ± desviación
- resultados comparados
- análisis del crecimiento del tiempo con más POIs

10. Informe final
Pregunta típica:
- “¿Qué te ha costado más?”
- “¿Qué decisiones de modelado tomaste?”
- “¿Qué limitaciones tiene tu solución?”

Qué debes tener:
- justificación del modelo
- decisiones de diseño
- limitaciones reales
- resultados experimentales
- conclusión técnica

La regla de oro:
Si tú eres capaz de responder estas 10 preguntas en 1 minuto sin mirar código, entonces estás listo.

Si quieres, te hago ahora mismo un “guion de defensa” de 2 minutos para que lo hables tal cual en la presentación.
